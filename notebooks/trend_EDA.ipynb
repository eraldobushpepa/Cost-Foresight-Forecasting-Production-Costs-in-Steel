```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8")
sns.set_theme()
```

```python
#Load Dataset

df = pd.read_csv("toy_dataset.csv")
df['Date'] = pd.to_datetime(df['Date'])

df.head()
```

```python

#SECTION 1 — Time Series Trend Analysis
#L’obiettivo è capire come si comportano nel tempo le variabili principali della produzione dell’acciaio.

#Plot Target Trend: Cost Per Trend



plt.figure(figsize=(12,6))
plt.plot(df['Date'], df['Cost_Per_Ton'], marker='o')
plt.title("Trend Over Time: Cost Per Ton", fontsize=16)
plt.xlabel("Quarter")
plt.ylabel("Cost Per Ton ($)")
plt.grid(True)
plt.show()
```

```python
#Core Inputs (Scrap, Electricity, Natural Gas)

fig, axes = plt.subplots(3, 1, figsize=(12,12), sharex=True)

features = [
    ('Scrap_Price', 'Scrap Price (Primary Raw Material)'),
    ('Electricity_Price', 'Electricity Price (EAF Energy Input)'),
    ('Natural_Gas_Price', 'Natural Gas Price (DRI Input)')
]

for ax, (col, title) in zip(axes, features):
    ax.plot(df['Date'], df[col], linewidth=2)
    ax.set_title(title, fontsize=14)
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
#SECTION 2 — Relationships Between Inputs and Cost

#Scatter: Scrap vs Cost Per Ton
plt.figure(figsize=(8,6))
sns.regplot(data=df, x="Scrap_Price", y="Cost_Per_Ton", line_kws={'color':'red'})
plt.title("Scrap Price vs Cost Per Ton")
plt.grid(True)
plt.show()

```

```python
#Scatter: Electricity vs Cost Per Ton

plt.figure(figsize=(8,6))
sns.regplot(data=df, x="Electricity_Price", y="Cost_Per_Ton", line_kws={'color':'red'})
plt.title("Electricity Price vs Cost Per Ton")
plt.grid(True)
plt.show()
```

```python
#Scatter: Natural Gas vs Cost Per Ton

plt.figure(figsize=(8,6))
sns.regplot(data=df, x="Natural_Gas_Price", y="Cost_Per_Ton", line_kws={'color':'red'})
plt.title("Natural Gas Price vs Cost Per Ton")
plt.grid(True)
plt.show()
```

```python
#SECTION 3 — Gross Margin Analysis (Advanced Business Insight)
#Gross Margin Per Ton Trend

plt.figure(figsize=(12,6))
plt.plot(df['Date'], df['Gross_Margin_Per_Ton'], marker='o', color='green')
plt.title("Gross Margin Per Ton Over Time")
plt.xlabel("Quarter")
plt.ylabel("Margin ($)")
plt.grid(True)
plt.show()
```

```python
#SECTION 4 — Advanced / Creative Features Trends
#Plot di 4 Features Avanzate Insieme

fig, axes = plt.subplots(2, 2, figsize=(14,10))
adv = [
    ('Diesel_Price', 'Diesel (Logistics)'),
    ('US_Dollar_Index', 'US Dollar Index'),
    ('Economic_Policy_Uncertainty_Index', 'EPU Index'),
    ('Graphite_Electrode_Price', 'Graphite Electrode Price')
]

for ax, (col, title) in zip(axes.flatten(), adv):
    ax.plot(df['Date'], df[col])
    ax.set_title(title, fontsize=12)
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
#SECTION 5 — Correlation Heatmap (Model Insight)
#Correlation Matrix

corr = df.drop(columns=["Date"]).corr()

plt.figure(figsize=(15,12))
sns.heatmap(corr, cmap="coolwarm", annot=False)
plt.title("Correlation Matrix — Quarterly Dataset")
plt.show()
```

```python
#SECTION 6 — Cost Breakdown Over Time

plt.figure(figsize=(14,7))
plt.stackplot(
    df['Date'],
    df['Scrap_Price'],
    df['Electricity_Price'],
    df['Natural_Gas_Price'],
    labels=['Scrap', 'Electricity', 'Natural Gas'],
    alpha=0.8
)

plt.title("Relative Contribution of Raw Inputs Over Time", fontsize=16)
plt.legend(loc='upper left')
plt.show()
```
